## 1. My lane (or freestyle) and why

**Lane: Refresh / Content Opportunity Scoring.** I chose it because "which page should we refresh next?" is a decision a content team makes every week, it is answerable from the observable signals in this dataset (impressions, clicks, position, engagement, freshness), and it produces a ranked queue — a concrete artifact a human can review. Compared to the other lanes it needs no private text or client data, so the whole exercise stays public-safe.

## 2. The question: decision, action, cost of a wrong call

**Decision:** which content items should a content team review for refresh, in what order?
**Who acts:** SEO/content operators triaging a refresh queue.
**Action:** the queue tells them the first 20 pages to read and what fix to test (refresh copy, improve CTR, expand to cover a secondary intent).
**Cost of a wrong call:** a page with high demand that is quietly declining gets fixed too late (lost visits), or a healthy page gets touched and regresses. Both are survivable but each wasted refresh costs a human's review time; the goal is to rank by *declining-with-demand* so the highest-leverage pages surface first.

## 3. Quick look at the data (2-3 real numbers)

Below I load the starter CSV and print three numbers that make this lane worth ~7 weeks: (1) how many rows are declining, (2) how many of those still carry search demand/impressions, and (3) the spread in taste of that opportunity.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("rows:", len(df))

declining = df["trend_direction"].eq("down").sum()
print("declining pages:", declining, f"({declining/len(df):.1%})")

# a *refreshable* page = declining but still receiving impressions (demand is there, content is underperforming)
candidates = df[(df["trend_direction"] == "down") & (df["impressions_90d"] > 0)]
print("declining WITH impressions (refresh candidates):", len(candidates), f"({len(candidates)/declining:.1%} of declining)")

# how concentrated is the opportunity? impressions distribution of candidates
print("candidate impressions_90d median:", int(candidates["impressions_90d"].median()),
      "· p90:", int(candidates["impressions_90d"].quantile(.9)))
```

## 4. Careful words: what I can and can't claim

This work can claim: *observed* associations between signal columns and a *trend label* that ships in the data; a transparent rule and a model that rank pages for **review**; and decision-support wording only.

This work can never claim: causality (that refreshing causes recovery), forecasting Google, 'ground-truth labelling' of the trend label, or producing any client-identifiable insights. All numbers are interpretable as *directional evidence for triage*, not proof of what will work.
